```python
tree = etree.parse(finding_aid_path)
print(tree)
```

Below is how to mark something as the root element

```python
root = tree.getroot()

print(root[:250])
```

XML in python uses qualified names "q names" -- each element is names according to its namespace and its element tag.
qname = namespace {uri} element tag

With this you can look through it like a list or dictionary

```python
for element in root: #if you want to refer to something in the whole document you can se root becase the above stuff set it up.
    print(element)
```

Asking for specific components of each element:
tag names, attributes, attached text

```python
for element in root:
    print(type(element), element.tag, element.text)
```

```python
for element in root:
    print(element.tag, element.attrib)
```

```python
for attribute in element.attrib:
    print(attribute)
```
