```python
import json
import requests
import csv
from pathlib import Path

base_url = "https://www.loc.gov/free-to-use/libraries/"
params = {
    'fo' : 'json'
}

r = requests.get(base_url, params=params) # Now I can access the underlying json data and treat it like a dictionary!
print(r.status_code)
```

```python
print(r.headers)
```

Look at the response (`r`) as json... and see if you can find what you are looking for (the list of items in the set):

```python
# assign the response to a variable (response) and use the built in .json() function of requests with r.json()
response = r.json()

# isolate the part of the response that you want, the item list, by using the named keys:
items = response['content']['set']['items']

# see if you have retrieved what you're looking for
for item in items:
    print(item)
```
