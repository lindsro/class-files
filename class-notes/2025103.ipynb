```python
import csv
import json
import requests

```

using requests to get data from loc

```python
base_url = 'https://www.loc.gov'
endpoint = '/search'

print('query url:', base_url + endpoint)
```

```python
params = {
    'fo' : 'json',
    'q' : 'yarn'
}

r = requests.get(base_url + endpoint, params=params)
print (r.status_code)
```

```python
for header in r.headers:
    print(header)
```
